```python
# Swiggy Order Details

swiggy_order = {
    "order_id": 101,
    "customer_name": "Farhana",
    "restaurant": "Paradise Biryani",
    "items": [
        {
            "item_name": "Chicken Biryani",
            "price": 250,
            "quantity": 2
        },
        {
            "item_name": "Coke",
            "price": 50,
            "quantity": 1
        }
    ],
    "delivery_partner": "Rahul",
    "delivery_time": "30 mins",
    "payment_mode": "UPI"
}
```

```python
print(swiggy_order["customer_name"])
```

```python
print(swiggy_order["items"][0]["item_name"])
```

```python
print(swiggy_order["items"][0]["price"])
```

```python
for item in swiggy_order["items"]:
    print(item["item_name"])
```

```python
#to access all the items
for item in swiggy_order["items"]:
    print("Item Name :", item["item_name"])
    print("Price :", item["price"])
    print("Quantity :", item["quantity"])
    print()
```

```python
total = 0

for item in swiggy_order["items"]:
    total += item["price"] * item["quantity"]

print("Total Amount =", total)
```
